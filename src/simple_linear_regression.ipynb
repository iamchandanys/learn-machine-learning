### Correlation

### A correlation matrix shows how closely related two or more variables are to each other using numbers called correlation coefficients.

- **1: Perfect positive correlation**  
  Two variables move exactly in the same direction. When one variable goes up, the other also goes up perfectly.

- **0: No correlation**  
  No relationship between the two variables. Changes in one variable do not predict changes in the other.

- **-1: Perfect negative correlation**  
  Two variables move exactly in opposite directions. When one variable goes up, the other goes down perfectly.

**In short:**  
- **1 = variables increase together**  
- **0 = variables unrelated**  
- **-1 = one increases while the other decreases**

## Simple Linear Regression

```python
# Divide the dataset into independent and dependent variables
# The independent variable (X) is typically the feature(s) used for prediction.
# The dependent variable (Y) is the target variable we want to predict.
# Here, we assume 'Weight' is the independent variable and 'Height' is the dependent variable.

# The independent variable should always be a 2D array or DataFrame Type.
X = df[['Weight']]
print(type(X))  # Output: <class 'pandas.core.frame.DataFrame'>
print("Independent Variable:", np.array(X).shape) # Output: (23, 1)  # 23 rows and 1 column

# The dependent variable can be a 1D array or Series Type.
Y = df['Height']
print(type(Y))  # Output: <class 'pandas.core.series.Series'>
print("Dependent Variable:", np.array(Y).shape)  # Output: (23,)  # 23 rows, no columns (1D array)
```

### Train Test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=42)

print("Training set size:", X_train.shape)  # Output 2D: (17, 1)
print("Testing set size:", X_test.shape)    # Output 2D: (6, 1)
```

### Train Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression(n_jobs=-1)  # Initialize the linear regression model with all available CPU cores
model.fit(X_train, Y_train) # Fit the model to the training data by passing training data
# print coefficients
print("Intercept (y-intercept):", model.intercept_)  # Output: Intercept (y-intercept) - 80.52684512622542
print("Coefficient (slope):", model.coef_)  # Output: Coefficient (slope) - [1.04792503]

# Intercept: The expected height when weight is zero (not always meaningful in real-world context).
# For example, if the intercept is 80.52684512622542, it means that if a person had zero weight (hypothetical), their height would be approximately 80.53 centimeters.
# Coefficient: With one unit increase in weight, how much does height increase? This is the slope of the line.
# For example, if the coefficient is 1.04792503, it means that for every additional kilogram of weight, the height increases by approximately 1.05 centimeters.

# Why only one coefficient?
# - In this case, we have only one independent variable (weight).

# Formula: ŷ = β0 + β1 * xi. β0 = intercept, β1 = coefficient[0], xi = independent variable (weight).
```

```python
# Predict on the test set
y_pred = model.predict(X_test)
print(y_pred)
print("Actual values:", Y_test.values)
```

### Plot Best Fit Line

```python
plt.scatter(X_train, Y_train, color='blue', label='Training Data')
plt.plot(X_train, model.predict(X_train), color='red', label='Regression Line')
```

### Predict the test data

```python
plt.scatter(X_test, Y_test, color='green', label='Test Data')
plt.plot(X_test, model.predict(X_test), color='orange', label='Predicted Line')
```

### Combine the training and test data for visualization

```python
plt.scatter(X_train, Y_train, color='blue', label='Training Data')
plt.plot(X_train, model.predict(X_train), color='red', label='Regression Line (Train)')

y_predict = model.predict(X_test)  # Predict the height for the test data
plt.scatter(X_test, Y_test, color='green', label='Test Data')
plt.plot(X_test, y_pred, color='orange', label='Regression Line (Test)')

plt.xlabel('Weight')
plt.ylabel('Height')
plt.title('Height vs Weight: Train & Test Data with Regression Lines')
plt.legend()
plt.grid(True)
plt.show()
```

### Errors

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Mean Squared Error (MSE): Average of squared differences between actual and predicted values.
mse = mean_squared_error(Y_test, y_predict)

# Root Mean Squared Error (RMSE): Square root of the MSE, easier to interpret in the same units as the target.
rmse = np.sqrt(mse)

# Mean Absolute Error (MAE): Average of absolute differences between actual and predicted values.
mae = mean_absolute_error(Y_test, y_predict)

print("Actual values:", Y_test.values.astype(int))
print("Predicted values:", y_predict.astype(int))
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("Mean Absolute Error (MAE):", mae)
```

### Performance Metrics

```python
from sklearn.metrics import r2_score

# R-squared (R²): Proportion of variance in the target explained by the model (ranges from 0 to 1; higher is better)
r2_squared = r2_score(Y_test, y_predict)

# Adjusted R-squared: Adjusts R² for the number of predictors and data points
n = X_test.shape[0]  # number of samples
p = X_test.shape[1]  # number of predictors
adjusted_r2 = 1 - (1 - r2_squared) * (n - 1) / (n - p - 1)

print("R-squared (R²):", r2_squared)
print("Adjusted R-squared:", adjusted_r2)
```
